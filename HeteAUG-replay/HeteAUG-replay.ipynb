```python
#After supplementing the fields used for feature extraction in the Kitsune dataset, perform traffic replay.
from scapy.all import *
import os

def generate_random_string(byte_size):
    random_bytes = os.urandom(byte_size)
    return random_bytes.hex()


random_string = generate_random_string(10)
print(f"Random string (10 bytes): {random_string}")
```

```python
from scapy.all import Ether, IP, TCP, wrpcap

packets = []
for i in range(1861):
    src_port = 12345 + i
    dst_port = 80
    seq_num = 1000 + i
    ack_num = 2000 + i
    payload = generate_random_string(int(2450/2))

    packet = Ether()/IP(src="192.168.2.1",dst="192.168.2.7")/TCP(sport=src_port, dport=8080, seq=seq_num, ack=ack_num)/payload
    packet.time = 1500802976.209211
    packet[Ether].src = "18:db:f2:01:09:33"
    packet[Ether].dst = "ff:ff:ff:ff:ff:ff" 
    packets.append(packet)


wrpcap("HeteAUG_replay.pcap", packets)
```
